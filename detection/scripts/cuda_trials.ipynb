```python
import torch

if torch.cuda.is_available():
    print(f"PyTorch está usando CUDA: {torch.version.cuda}")
    # Comprobar si cuDNN está habilitado y la versión que PyTorch está usando
    print(f"cuDNN habilitado: {torch.backends.cudnn.enabled}")
    if torch.backends.cudnn.enabled:
        # Esta línea te da la versión de cuDNN que PyTorch detecta y está usando
        print(f"Versión de cuDNN (detectada por PyTorch): {torch.backends.cudnn.version()}")
else:
    print("CUDA no está disponible para PyTorch.")
```
